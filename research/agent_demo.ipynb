```python
import os
import certifi


from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_community.tools.tavily_search import TavilySearchResults
from langchain import hub
from langchain.tools import tool
```

```python
from langchain.agents import create_react_agent, AgentExecutor
```

```python
#load environment variables from .env file

os.environ["SSL_CERT_FILE"] = certifi.where()
load_dotenv()

GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")
TAVILY_API_KEY = os.getenv("TAVILY_API_KEY")
WEATHERSTACK_API_KEY= os.getenv("WEATHERSTACK_API_KEY")
```

```python
search_tool = TavilySearchResults(max_results=3)
```

```python
#initialize the ChatGroq model with the specified parameters

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    google_api_key=os.getenv("GOOGLE_API_KEY")
)
```

```python
response = llm.invoke("what year is it?")
response
```

```python
from langsmith import Client

client = Client()

prompt = client.pull_prompt("hwchase17/react")
```

```python
tools = [search_tool]
```

```python
# create agent

agent = create_react_agent(
    llm=llm,
    tools=tools,
    prompt=prompt
)
```

```python
# agent executor

agent_executor = AgentExecutor(
    agent=agent,
    tools=tools,
    verbose=True # we can see logs 
)
```

```python

# run agent executor with a query

response = agent_executor.invoke({
    "input": (
        "Tell me the lastest news about Usa and Iran war."  
    )
})


```
